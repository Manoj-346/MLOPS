```python
TARGET_CANDIDATES = [
    "Sales Amount",
    "SalesAmount",
    "Sales_Amount",
    "Sales",
    "sales",
    "Salary",
    "salary",
    "Employee Salary",
    "Employee_Salary",
    "Amount",
    "amount",
    "Revenue",
    "revenue"
]

def find_target_column(columns):
    normalized = {
        str(col).strip().lower().replace("_", " ").replace("-", " "): col
        for col in columns
    }

    for candidate in TARGET_CANDIDATES:
        key = candidate.lower().replace("_", " ").replace("-", " ")
        if key in normalized:
            return normalized[key]

    return None


TARGET_COLUMN = find_target_column(df.columns)

print("Target column detected:", TARGET_COLUMN)

if TARGET_COLUMN is None:
    print("\nCould not automatically detect the target.")
    print("Available columns:")
    print(list(df.columns))
    print("\nSet TARGET_COLUMN manually, for example:")
    print('TARGET_COLUMN = "Sales Amount"')
else:
    print("Target column:", TARGET_COLUMN)
```

```python
#Data cleaning
print("=" * 70)
print("DATA ENGINEERING / CLEANING")
print("=" * 70)

data = df.copy()

# Remove completely empty rows
data = data.dropna(how="all")

# Remove duplicate rows
before_duplicates = len(data)
data = data.drop_duplicates()
after_duplicates = len(data)

# Convert target to numeric
data[TARGET_COLUMN] = pd.to_numeric(
    data[TARGET_COLUMN],
    errors="coerce"
)

# Remove rows where target is missing
before_target = len(data)
data = data.dropna(subset=[TARGET_COLUMN])
after_target = len(data)

print("Original rows:", len(df))
print("Removed duplicates:", before_duplicates - after_duplicates)
print("Removed missing target rows:", before_target - after_target)
print("Final rows:", len(data))

display(data.head(10))
```

```python
#Feature engineering
print("=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)

feature_data = data.copy()

# Remove target from features
X = feature_data.drop(columns=[TARGET_COLUMN])
y = feature_data[TARGET_COLUMN]

print("Target:", TARGET_COLUMN)

print("\nFeature columns:")
for col in X.columns:
    print("•", col)

print("\nTarget statistics:")
display(y.describe().to_frame("Target"))

print("\nFeature dataset:")
display(X.head())
```

```python
#Determine numeric/categorical columns
numeric_features = X.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("=" * 70)
print("FEATURE ANALYSIS")
print("=" * 70)

print("\nNumeric features:")
for col in numeric_features:
    print("•", col)

print("\nCategorical features:")
for col in categorical_features:
    print("•", col)
```

```python
#Train/Test split
from sklearn.model_selection import train_test_split

TEST_SIZE = 0.20
RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

print("=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
print("Test size    :", TEST_SIZE)
print("Random state :", RANDOM_STATE)
```

```python
#Preprocessing pipeline
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

print("✓ Preprocessing pipeline created.")
```

```python
#Import models
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=RANDOM_STATE
    )
}

print("Models to train:\n")

for name in models:
    print("•", name)
```

```python
# ============================================================
# CELL 21 - MODEL TRAINING + MLFLOW TRACKING
# CORRECTED FOR MLFLOW 3.15.2
# ============================================================

import mlflow
import mlflow.sklearn

import numpy as np
import pandas as pd
import joblib

from pathlib import Path
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# 1. CONNECT TO EXISTING MLFLOW SERVER
# ============================================================

mlflow.set_tracking_uri(
    "http://127.0.0.1:5000"
)

mlflow.set_experiment(
    "Sales_Prediction"
)

print("=" * 80)
print("CELL 21 - MODEL TRAINING + MLFLOW TRACKING")
print("=" * 80)

print("\nMLflow Tracking URI:")
print(mlflow.get_tracking_uri())

# ============================================================
# 2. CREATE REQUIRED FOLDERS
# ============================================================

Path("models").mkdir(exist_ok=True)
Path("reports").mkdir(exist_ok=True)

# ============================================================
# 3. RESULTS STORAGE
# ============================================================

results = []

# ============================================================
# 4. TRAIN EACH MODEL
# ============================================================

for model_name, original_model in models.items():

    print("\n")
    print("=" * 80)
    print(f"TRAINING: {model_name}")
    print("=" * 80)

    # Create fresh model
    model = clone(original_model)

    # Create fresh preprocessing pipeline
    current_preprocessor = clone(preprocessor)

    # Complete ML pipeline
    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                current_preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    # ========================================================
    # START MLFLOW RUN
    # ========================================================

    with mlflow.start_run(
        run_name=model_name.replace(" ", "_")
    ):

        # ----------------------------------------------------
        # PARAMETERS
        # ----------------------------------------------------

        mlflow.log_param(
            "model",
            model.__class__.__name__
        )

        mlflow.log_param(
            "test_size",
            TEST_SIZE
        )

        mlflow.log_param(
            "random_state",
            RANDOM_STATE
        )

        if hasattr(model, "n_estimators"):
            mlflow.log_param(
                "n_estimators",
                model.n_estimators
            )

        if hasattr(model, "max_depth"):
            mlflow.log_param(
                "max_depth",
                model.max_depth
            )

        # ----------------------------------------------------
        # TAGS
        # ----------------------------------------------------

        mlflow.set_tag(
            "model_name",
            model_name
        )

        mlflow.set_tag(
            "problem_type",
            "regression"
        )

        mlflow.set_tag(
            "target_column",
            TARGET_COLUMN
        )

        # ----------------------------------------------------
        # TRAIN
        # ----------------------------------------------------

        print("Training model...")

        pipeline.fit(
            X_train,
            y_train
        )

        print("✓ Training completed")

        # ----------------------------------------------------
        # PREDICTION
        # ----------------------------------------------------

        y_pred = pipeline.predict(
            X_test
        )

        print("✓ Prediction completed")

        # ----------------------------------------------------
        # CALCULATE METRICS
        # ----------------------------------------------------

        mae = mean_absolute_error(
            y_test,
            y_pred
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_test,
                y_pred
            )
        )

        r2 = r2_score(
            y_test,
            y_pred
        )

        print("\nMetrics:")
        print(f"MAE  : {mae:.4f}")
        print(f"RMSE : {rmse:.4f}")
        print(f"R²   : {r2:.4f}")

        # ----------------------------------------------------
        # LOG METRICS
        # ----------------------------------------------------

        mlflow.log_metric(
            "MAE",
            float(mae)
        )

        mlflow.log_metric(
            "RMSE",
            float(rmse)
        )

        mlflow.log_metric(
            "R2",
            float(r2)
        )

        print("✓ Metrics logged")

        # ----------------------------------------------------
        # SAVE MODEL LOCALLY
        # ----------------------------------------------------

        model_filename = (
            model_name
            .lower()
            .replace(" ", "_")
            + ".pkl"
        )

        model_path = (
            Path("models") /
            model_filename
        )

        joblib.dump(
            pipeline,
            model_path
        )

        print(
            "✓ Local model saved:",
            model_path
        )

        # ----------------------------------------------------
        # LOG MODEL FILE AS ARTIFACT
        # ----------------------------------------------------

        mlflow.log_artifact(
            str(model_path),
            artifact_path="model_files"
        )

        print("✓ Model artifact logged")

        # ----------------------------------------------------
        # LOG MODEL TO MLFLOW
        #
        # IMPORTANT:
        # Explicitly use pickle to avoid the skops
        # numpy.dtype trust error.
        # ----------------------------------------------------

        print("Logging model to MLflow...")

        mlflow.sklearn.log_model(
            sk_model=pipeline,
            name="model",
            serialization_format="pickle"
        )

        print("✓ Model successfully logged to MLflow")

        # ----------------------------------------------------
        # GET RUN ID
        # ----------------------------------------------------

        run_id = (
            mlflow.active_run()
            .info
            .run_id
        )

        print("Run ID:", run_id)

        # ----------------------------------------------------
        # SAVE RESULT
        # ----------------------------------------------------

        results.append({
            "Run ID": run_id,
            "Model": model_name,
            "MAE": float(mae),
            "RMSE": float(rmse),
            "R2": float(r2)
        })

        print("✓ RUN COMPLETED SUCCESSFULLY")


# ============================================================
# 5. CREATE RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    results
)

print("\n")
print("=" * 80)
print("ALL MODELS COMPLETED SUCCESSFULLY")
print("=" * 80)

display(
    results_df
)
```
